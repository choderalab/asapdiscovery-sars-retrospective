```python
import itertools
import numpy as np
import pandas as pd
```

```python
refs = ['PDB123', 'PDB456']
ligs = ['LIG_A', 'LIG_B', 'LIG_C',]
ref_lig_combinations = [(ref, lig) for (ref, lig) in itertools.product(refs, ligs)]
ref_data = pd.DataFrame({"Reference_Structure": refs, "Ref_Data_1": [np.random.random() for ref in refs]})
query_data = pd.DataFrame({"Query_Ligand": ligs, "Lig_Data_1": [np.random.random() for lig in ligs]})
pose_results = pd.DataFrame.from_records([{"Reference_Structure": ref, 
                                  "Query_Ligand": lig, 
                                  "RMSD": np.random.random()*8,
                                 "Pose_ID": pose,
                                           "Method": method
                                 } 
                                 for ref, lig, pose, method in itertools.product(refs, ligs, range(0,2), ["FRED", "POSIT"])])
ecfp = pd.DataFrame.from_records([{"Reference_Structure": ref, 
                                   "Query_Ligand": lig, 
                                   "Type": "ECFP", 
                                   "Tanimoto": np.random.random()} 
                                  for ref, lig in ref_lig_combinations])

tanimotocombo = pd.DataFrame.from_records([{"Reference_Structure": ref, 
                                            "Query_Ligand": lig, 
                                            "Type": "TanimotoCombo", 
                                            "Tanimoto": np.random.random(), 
                                            "Aligned": aligned} 
                                           for ref, lig, aligned in itertools.product(refs, ligs, ["True", "False"])])
```

```python
dfs = [ref_data, query_data, pose_results, ecfp, tanimotocombo]
```

```python
total_cells = sum([df.size for df in dfs])
```

```python
total_cells
```

# combine pose data

```python
combined_pose_data = pose_results.merge(ref_data, on=['Reference_Structure'])
combined_pose_data = combined_pose_data.merge(query_data, on=['Query_Ligand'])
```

# simple merging changes the column names

```python
from functools import reduce
# add pose information to pose data
simple_combination = reduce(lambda x, y: x.merge(y, on=['Reference_Structure', 'Query_Ligand']), [combined_pose_data, ecfp, tanimotocombo])
```

```python
simple_combination
```

```python
simple_combination.size
```

```python
print(f"size increase is {simple_combination.size / total_cells:.2f}x")
```

# Combine similarities first

```python
similarities = pd.concat([ecfp, tanimotocombo])
```

```python
similarities
```

## then combine

```python
big_combination = combined_pose_data.merge(similarities, on=['Reference_Structure', 'Query_Ligand'])
```

```python
big_combination
```

```python
print(f"size increase is {big_combination.size / total_cells:.2f}x")
```
