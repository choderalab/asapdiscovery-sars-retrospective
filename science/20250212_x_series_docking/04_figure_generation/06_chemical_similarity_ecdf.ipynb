# the purpose of this notebook is to generate an eCDF plot of the chemical similarity of the compounds in the dataset

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from pathlib import Path
import plotly.express as px
import plotly.graph_objects as go
import numpy as np
```

```python
figure_path = Path("figures")
figure_tag = Path("06_chemical_similarity_ecdf")
```

```python
def get_svg(fig_name) -> Path:
    return figure_path / f"{figure_tag}_{fig_name}.svg"

def get_png(fig_name) -> Path:
    return figure_path / f"{figure_tag}_{fig_name}.png"

def save_both(fig_name):
    plt.savefig(get_svg(fig_name), format='svg', bbox_inches='tight')
    plt.savefig(get_png(fig_name), format='png', bbox_inches='tight')

def save_both(fig, fig_name=""):
    fig.write_image(get_svg(fig_name))
    fig.write_image(get_png(fig_name))
```

```python
# I'll end up want to use different data, I'd like to convert this into a python script
# raw_df = pd.read_csv("/Users/alexpayne/Scientific_Projects/mers-drug-discovery/sars2-retrospective-analysis/20250212_p_to_x_posit/20250311_combined_results") 
similarity_data = pd.read_csv("/Users/alexpayne/Scientific_Projects/asapdiscovery-sars-retrospective/data/all_tanimoto.csv", index_col=0)
```

# make ecdf plot

```python
fig_name = "v1"
# Get unique similarity metrics
similarity_metrics = similarity_data['Similarity Metric'].unique()

# Create a new figure
fig = go.Figure()

# Add an ECDF trace for each similarity metric
for metric in similarity_metrics:
    # Filter data for this metric
    metric_data = similarity_data[similarity_data['Similarity Metric'] == metric]['Tanimoto']
    
    # Sort the data and compute ECDF
    x = np.sort(metric_data)
    y = np.arange(1, len(x) + 1) / len(x)
    
    # Add trace
    fig.add_trace(go.Scatter(
        x=x,
        y=y,
        mode='lines',
        name=metric,
        line=dict(width=3)
    ))

large_font = 24
small_font = 18
tick_font=18

fig.update_layout(
    # Figure dimensions
    height=550,
    width=750,
    template="simple_white",
    margin=dict(l=0, r=20, t=0, b=0),
    
    # Font settings
    font_family="Helvetica",
    
    # Legend settings
    legend_title_text="<b> Similarity Metric </b>",
    legend=dict(yanchor="bottom", y=0, xanchor="right", x=1, font=dict(size=small_font, family='Helvetica'), title=dict(font=dict(size=small_font, family='Helvetica'))),
    
    # X-axis settings
    xaxis=dict(
        title_text="<b> Tanimoto Coefficient </b>",
        title_font=dict(size=large_font),
        color="black",
        tickfont=dict(size=tick_font, family="Helvetica", color="black"),
        tickmode="auto",  # Can be "auto", "linear", "array"
        nticks=10,        # Approximate number of ticks when using "auto" mode
        tickangle=0,      # Rotation angle of tick labels
        ticklen=8,        # Length of tick marks
        tickwidth=2,      # Width of tick marks
        showgrid=False,    # Show grid lines
        gridcolor="lightgray",  # Color of grid lines
        gridwidth=1,      # Width of grid lines
        range=(0,1)
    ),
    
    # Y-axis settings
    yaxis=dict(
        title_text="<b> Fraction of Pairwise Ligand Similarities </b>",
        title_font=dict(size=large_font),
        color="black",
        tickfont=dict(size=tick_font, family="Helvetica", color="black"),
        tickmode="auto",
        nticks=8,
        tickformat=".1f",  # Format for tick labels (1 decimal place)
        ticklen=8,
        tickwidth=2,
        showgrid=False,
        gridcolor="lightgray",
        gridwidth=1,
        range=(0,1),  # Commented out as in original
    ),
)

fig.show()
```

```python
save_both(fig, fig_name)
```
