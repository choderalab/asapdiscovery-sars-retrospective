```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from pathlib import Path
import plotly.express as px
import plotly.graph_objects as go
import numpy as np
```

```python
figure_path = Path("figures")
figure_tag = Path("09_dataset_split")
```

```python
def get_svg(fig_name) -> Path:
    return figure_path / f"{figure_tag}_{fig_name}.svg"

def get_png(fig_name) -> Path:
    return figure_path / f"{figure_tag}_{fig_name}.png"

def save_both(fig_name):
    plt.savefig(get_svg(fig_name), format='svg', bbox_inches='tight')
    plt.savefig(get_png(fig_name), format='png', bbox_inches='tight')

# def save_both(fig, fig_name=""):
#     fig.write_image(get_svg(fig_name))
#     fig.write_image(get_png(fig_name))
```

```python
# I'll end up want to use different data, I'd like to convert this into a python script
raw_df = pd.read_csv("/data1/choderaj/paynea/asap-datasets/20250212_p_to_x_posit/analyzed_data/results.csv", index_col=0)
```

```python
raw_df.columns
```

```python
from matplotlib.ticker import ScalarFormatter
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))
fig = sns.lineplot(data=raw_df,
                  x="N_Per_Split",
                  y="Fraction",
                  hue="Score",
                  style="Split",
                  palette='crest')

# Set font sizes
title_font_size = 24
text_font_size = 18

# Add labels with bold font and specified sizes
plt.xlabel('Number of Reference Structures Available to Use', 
          fontsize=title_font_size, 
          fontweight='bold',
          )
plt.ylabel('Fraction of Ligands\nPosed <2Å from Reference', 
          fontsize=title_font_size, 
          fontweight='bold',
          )

# Set x-axis to log scale but show raw numbers
plt.xscale('log')
plt.gca().xaxis.set_major_formatter(ScalarFormatter())

# Set custom tick locations with specified font size
custom_ticks = [1, 5, 10, 20, 50, 100, 200, raw_df["N_Per_Split"].max()]
plt.xticks(custom_ticks, custom_ticks, fontsize=text_font_size)
plt.yticks(fontsize=text_font_size)

# Update legend font size
legend = plt.legend()
plt.setp(legend.get_title(), fontsize=title_font_size, fontweight='bold')
plt.setp(legend.get_texts(), fontsize=text_font_size)

plt.tight_layout()

fig_name = "comparison"
plt.savefig(get_svg(fig_name), format='svg', bbox_inches='tight')
plt.savefig(get_png(fig_name), format='png', bbox_inches='tight')
```

```python
from matplotlib.ticker import ScalarFormatter
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))

# First create the main plot to get the color mapping
fig = sns.lineplot(data=raw_df,
                  x="N_Per_Split",
                  y="Fraction",
                  hue="Score",
                  style="Split",
                  palette='crest')

# Get the color mapping
legend = plt.gca().get_legend()
color_map = {text._text: line.get_color() 
            for line, text in zip(legend.get_lines(), legend.get_texts())}

# Clear the plot to start fresh
plt.clf()
plt.figure(figsize=(12, 8))

# Create fill between for each group using matched colors
for name, group in raw_df.groupby(['Score', 'Split']):
    score = name[0]  # First element is Score
    plt.fill_between(group['N_Per_Split'],
                    group['CI_Lower'],
                    group['CI_Upper'],
                    color=color_map[score],
                    alpha=0.2)

# Recreate the main plot
fig = sns.lineplot(data=raw_df,
                  x="N_Per_Split",
                  y="Fraction",
                  hue="Score",
                  style="Split",
                  palette='crest')

plt.xlabel('Number of Reference Structures Available to Use',
          fontsize=24,
          fontweight='bold')
plt.ylabel('Fraction of Ligands\nPosed <2Å from Reference',
          fontsize=24,
          fontweight='bold')

plt.xscale('log')
plt.gca().xaxis.set_major_formatter(ScalarFormatter())

custom_ticks = [1, 5, 10, 20, 50, 100, 200, raw_df["N_Per_Split"].max()]
plt.xticks(custom_ticks, custom_ticks, fontsize=18)
plt.yticks(fontsize=18)

legend = plt.legend()
plt.setp(legend.get_title(), fontsize=24, fontweight='bold')
plt.setp(legend.get_texts(), fontsize=18)

plt.tight_layout()

fig_name = "comparison"
plt.savefig(get_svg(fig_name), format='svg', bbox_inches='tight')
plt.savefig(get_png(fig_name), format='png', bbox_inches='tight')
```
